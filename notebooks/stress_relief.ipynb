```python
# Problem - 3
# ADDING STESS AND RELIEF PERIOD USING NET-FLOW OF CBP AND HHS AND THE DAILY INTAKE OF BOTH OF THEM
import pandas as pd
from pathlib import Path
import numpy as np
```

```python
BASE_DIR = Path.cwd().parent
file_path = BASE_DIR / "data" / "cleaned_data" / "sorted_data.csv"

df = pd.read_csv(file_path, index_col=0)
```

```python
df.head()
```

```python
avg_load_on_CBP = df['total load on CBP'].mean()
print(avg_load_on_CBP)

max_CBP_Custody_net_change = df['CBP Custody net change'].max()
print(max_CBP_Custody_net_change)
```

```python
def classify(row):
    if ((row['total load on CBP'] > avg_load_on_CBP) & (row['CBP Custody net change'] > 20)):
        return "Stress"
    else:
        return "Relief"

df['daily status'] = df.apply(classify, axis=1)
```

```python
values = df['daily status'] == "Stress"
print(values.sum())
```
