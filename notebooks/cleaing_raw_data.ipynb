### Data Ingestion and Data Cleaning

```python
# importing pandas and os library
import pandas as pd
import os
```

```python
# loading the dataset to df: DataFrame
df = pd.read_csv(r'C:\Users\as330\OneDrive\Desktop\UAC-Care-System\data\raw\HHS_Unaccompanied_Alien_Children_Program.csv')
```

```python
df = df.dropna()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
```

```python
df.head()
```

```python
df =  df.sort_values(by="Date")
```

```python
df.head()
```

```python
df = df.set_index("Date")
```

```python
daily_index = pd.date_range(
    start=df.index.min(),
    end=df.index.max(),
    freq="D"
)
```

```python
df = df.reindex(daily_index)
```

```python
cols = ["Children in HHS Care"]

for col in cols:
    df[col] = pd.to_numeric(
        df[col].astype(str).str.replace(",", "", regex=False),
        errors="coerce"
    )
```

```python
df.shape
```

```python
import os

folder_path = r'UAC-Care-System\data\cleaned_data'

print("Path:", folder_path)
print("Parent exists:", os.path.exists(os.path.dirname(folder_path)))

os.makedirs(folder_path, exist_ok=True)

print("Folder exists:", os.path.exists(folder_path))
```

```python
df.to_csv(os.path.join(folder_path, 'cleaned_data.csv'), index=True)
```
