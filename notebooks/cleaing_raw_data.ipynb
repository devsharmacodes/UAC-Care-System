```python
data = data.dropna()
```

```python
data["Date"] = pd.to_datetime(data["Date"], errors="coerce")
```

```python
data.head()
```

```python
data =  data.sort_values(by="Date")
```

```python
data.head()
```

```python
data = data.set_index("Date")
```

```python
daily_index = pd.date_range(
    start=data.index.min(),
    end=data.index.max(),
    freq="D"
)
```

```python
data = data.reindex(daily_index)
```
